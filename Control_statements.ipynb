<pre>
    topic: if else
    
    Write a program to verify whether the given string is panagram or not
    Panagram: all 26 alphabets should be present in the string (Irrespective of their case)

    input: The quick brown fox jumps over the lazy dog
    output: yes
</pre>

```python
s=input()
s=s.lower()
l=[]
for i in s:
    if (i.isalpha() and (i not in l)):
        l.append(i)
if(len(l) == 26):
    print("Yes")
else:
    print("No")

```

```python
#Panagram in simple way

import string
s=input()
u=string.ascii_uppercase
l=string.ascii_lowercase

if all((i in s or j in s) for i,j in zip(u,l)):   #all() returns true if all cases are true
    print("Yes")
else:
    print("No")
```

```python
#Panagram in other way

import string
s=input().lower()
l=string.ascii_lowercase

if all((i in s ) for i in l):   #all() returns true if all cases are true
    print("Yes")
else:
    print("No")
```

<pre>
    topic:nested if
    
    Write a program to read 3 subjects marks of student, admission is confirmed for a student if he passed in all subjects(>=40) and score is greater than 64 in atleast 2 subjects.
</pre>

```python
s1,s2,s3=map(int,input().split())
if(s1>=40 and s2>=40 and s3>=40):
    if((s1>=65 and s2>=65) or (s2>=65 and s3>=65) or (s1>=65 and s3>=65)):
        print("admission is confirmed")
    else:
        print("admission is not confirmed")
else:
    print("fail")
```

<pre>
    topic: elif ladder
    
    Find whether given character is Upper, Lower, digit, Space or Puntuation.
</pre>

```python
c=input()
if c.isupper():
    print("Upper")
elif c.islower():
    print("Lower")
elif c.isdigit():
    print("digit")
elif c.isspace():
    print("Space")
else:
    print("Puntuation")
```
