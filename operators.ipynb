```python
#Binary operators
#1.Arithmetic Operators
print(5+10)
print(5-10)
print(5*10)
print(5/10)
print(5%10)
print(5//10)
print(5**2)

print(100.20 % 10.50)  # % works on float, int

print(pow(5,2))
```

```python
#Binary operators
#2.Comparision operators
#2.1.Relational Operators
print(5<10)
print(5<=10)
print(5>10)
print(5>=10)
print(5==10)
print(5!=10)

print("\n");
b=True
a=False

print(a<b)
print(a<=b)
print(a>b)
print(a>=b)
print(a==b)
print(a!=b)

print("\n");
#print("s1">2)  - TypeError: '>' not supported between instances of 'str' and 'int'
#print(3+4j>2)  - TypeError: '>' not supported between instances of 'complex' and 'int'
print(1.2>2)
print(1.2>True)
print(-1.2>False)
```

```python
#Binary operators
#2.Comparision operators
#2.2.Logical Operators
print(5 and 10)
print(5 or 10)
print(not 10)    # Unary operator
```

```python
a=0
b=2
if(a and b*2):    #lazy analyzer - b*2 is not happening
    print(a*2,b*2)
else:
    print(a+2,b+2)
```

```python
a=2
b=0
if(a or b+2):   #lazy analyzer - b+2 is not happening
    print(a*2,b*2)
else:
    print(a+2,b+2)
```

```python
#Binary operators
#3.Assignment operator (short hand operator)
#works with arithmetic and bitwise operators
a=5
b=2
a=a+b
print(a)

a+=b   #assignment with arithmetic
print(a)

print(5<<2)
print(5>>2)
print(~5)   # -(data+1)
print(~-5)  # -(data+1)
```
