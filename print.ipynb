```python
a=3214
print("%d"%(a))
print("%3d"%(a))
print("%9d"%(a))
print("%-9d"%(a),end="end\n")
print("%o"%(a))
print("%x"%(a))
```

```python
a=int(input())
b=bin(a)   #bin(17) - gives 0b10001 string
print(b,type(b))

b=b[2:]
print(b,type(b))

b=int(b,2)    #int(data,base) - converts to decimal
print(b,type(b))
```

```python
a=int(input())
b=oct(a)   #oct(36) - gives 0o44 string
print(b,type(b))

b=b[2:]
print(b,type(b))

b=int(b,8)    #int(data,base) - converts to decimal
print(b,type(b))
```

```python
a=int(input())
b=hex(a)   #hex(3214) - gives 0xc8e string
print(b,type(b))

b=b[2:]
print(b,type(b))

b=int(b,16)    #int(data,base) - converts to decimal
print(b,type(b))
```

```python
a=2500.321
print("%f"%(a))
print("%.f"%(a))
print("%.3f"%(a))
print("%.7f"%(a))
print("%3.2f"%(a))
print("%12.3f"%(a))
print("%-12.3f"%(a),end="end\n")
```
